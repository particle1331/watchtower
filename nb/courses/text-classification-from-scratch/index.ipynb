This course builds an end-to-end **text classification system** using AI-written-text detection as its main case study. It follows the [ai-detector-from-scratch repository](https://github.com/rasbt/ai-detector-from-scratch) from data construction through evaluation and local inference, and adds a historical path from bag-of-words models to neural sequence classifiers, transformers, and general-purpose decision APIs.

**Design rule:** treat every score as evidence about a model on a defined data distribution, not as proof about an individual text or a promise of generalization to new tasks.

By the end, you will have a reproducible detector report and, for the capstone, a local Jev-like decision system implemented around a pretrained ModernBERT encoder. You will train a shared candidate-scoring model across text-classification tasks, expose Choice, Noul, and Score operations, and measure how well it transfers to a task it did not train on. This is an educational implementation of an interface and scoring design described in the article; it is not a reproduction of Jev's proprietary model or training process.

## Prerequisites

**Tooling.** The core reading path needs Python 3.12, `uv`, Git, and a notebook environment. Start with the repository's lightweight logistic-regression path. Transformer inference and training add dependencies; the ModernBERT capstone is easiest with a CUDA GPU, though a small-data smoke run can use less capable hardware. Recreating the full detector corpus can require external model APIs or a local Ollama service. Those paths are optional.

**Knowledge.** Basic Python, arrays/data frames, train/validation/test splits, and binary classification are assumed. Familiarity with precision, recall, calibration, and logistic regression helps; Chapters 01 and 07 establish the key terms.

## Execution at a glance

For a quick local detector run, use the repository's published dataset and logistic-regression model artifact. Read each stage README before running it and record the artifact version. The detector's standard path can fine-tune selected transformer models using the published dataset. The capstone uses public task datasets to train one ModernBERT-backed scorer across multiple decision tasks; it does not require a hosted Jev account or API key. Full reproduction of the detector corpus starts with licensed human text and generates AI and AI-assisted text; that path uses substantially more compute and may make external API calls.

**Reading sources.** The [repository](https://github.com/rasbt/ai-detector-from-scratch) supplies the runnable detector pipeline. The article [Language Models for Text Classification: From Bag-of-Words to Jev](https://magazine.sebastianraschka.com/p/classifier-history-and-jev) supplies the architecture history and describes a shared one-score-per-candidate ModernBERT design. Jev's internal architecture, training data, and training method are not public in the article, so this capstone builds a transparent Jev-like system rather than claiming to recreate Jev.
