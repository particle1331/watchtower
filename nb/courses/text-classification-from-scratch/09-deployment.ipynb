A useful model needs a stable inference contract. The repository's reusable `src/ai_detector` package, CLI, and local API show how model artifacts can be downloaded and invoked without repeating training. The article's Choice, Noul, and Score interfaces provide a second way to think about structured classification inputs and outputs.

## Inference contract

For the repository detector, specify accepted input fields, maximum length, chunking behavior, output schema, positive-class direction, model identifier, and error handling. Its CLI returns a numeric score from 0 to 100, where larger values indicate stronger model support for the AI-generated class. That score is not proof about authorship. If the service emits a thresholded label, version the threshold separately from the model.

A structured general-purpose API adds task instructions and a response schema:

| Response form | Contract | Integration check |
|---|---|---|
| Choice | One selected option from named alternatives | Confirm options are mutually exclusive and map every returned label. |
| Noul | A yes/no probability for each question | Treat questions independently unless the API explicitly defines a joint distribution. |
| Score | A value on an ordered rubric | Define each rubric level and validate its ordinal meaning. |

Do not assume “confidence” means the winning class probability. Record the API's documented meaning, model/version, request parameters, and response shape. Validate output parsing, missing fields, timeouts, rate limits, retries, and repeated-call variation.

Load a local model once when a service starts, validate inputs, apply the same tokenizer and truncation rules used during evaluation, and return predictable JSON. For long documents, define how chunk scores are aggregated. For remote services, account for network latency, service availability, per-request cost, and whether sending user text is acceptable. Log model version and operational failures while minimizing text retention.

## Exercise: deployment check

Use the repository's minimal logistic-regression download and CLI path, or inspect its API code. Send a normal input, an empty input, a long input, and malformed input. Record output shape, errors, model identifier, and latency.

Then design a generic request/response contract for one Choice, Noul, or Score use case. Include validation and fallback behavior for unavailable or malformed responses. Draft a user-facing note that describes what the score means and what it cannot establish. Do not deploy a detector as sole evidence for a consequential judgment.
