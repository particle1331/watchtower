Instruction tuning changes the examples and objective after pretraining. This chapter will make the prompt-response format explicit and evaluate a fixed set of requests.

## Planned content

Study instruction, optional input, and response fields; padding and supervised-loss masking; and the difference between held-out loss and answer quality. Define a review rubric for compliance, correctness, and unsupported claims.

## Planned lab and evidence

Inspect formatted examples, fine-tune from a declared checkpoint, and compare before-and-after responses on fixed prompts. Record sampling settings and failure cases. Any optional automated grader must be named, with original responses retained. The response set becomes the second capstone branch.
