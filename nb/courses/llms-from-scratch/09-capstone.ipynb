The capstone will turn the chapter artifacts into an auditable account of what the small GPT learned and how downstream objectives changed its behavior.

## Planned deliverable

Produce a report and runnable configuration tracing data source, tokenizer, splits, model, training settings, checkpoints, and evaluation prompts. Include pretraining loss and fixed-prompt samples, classification metrics and inspected errors, and instruction response comparisons. Include LoRA only if Chapter 08 was completed.

## Acceptance checks

- A fresh smoke run reaches tokenization, forward pass, loss, update, save, and generation.
- Each checkpoint has a stated starting point and objective.
- Evaluation data and prompts are fixed before checkpoint comparisons.
- Local scratch training is distinguished from externally loaded weights.
- The report states compute and data limits and observed failure cases.

The claim is reproducibility and understanding of the pipeline. A small run is not evidence of broad language-model capability.
