Attention alone does not make a language model. This chapter will assemble the decoder so the text batches can produce vocabulary scores.

## Planned content

Explain residual connections, layer normalization, feed-forward networks, repeated decoder blocks, final normalization, and vocabulary projection. Relate context length and parameter count to memory needs, and explain why generation reads final-position scores.

## Planned lab and evidence

Implement a configurable small GPT, inspect intermediate shapes, and count parameters. Run a deterministic forward check and generate a few tokens from the untrained model to verify the interface. Confirm that the causal property survives composition. The generation-ready model passes to Chapter 05.
