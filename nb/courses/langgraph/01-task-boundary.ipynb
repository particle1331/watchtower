An ordinary search pipeline can answer many repository questions. This chapter identifies the point where a change investigation needs explicit state, parallel evidence branches, bounded refinement, review, or resume. The comparison uses the same pinned repository fixture throughout the course.


## The task-boundary test

Represent a task by six operational signals. The threshold is a decision aid, not a framework slogan: a graph deserves evaluation only when its control surface makes a material workflow contract visible.


```python
from dataclasses import dataclass
from IPython.display import Markdown, display

@dataclass(frozen=True)
class TaskProfile:
    name: str
    typed_handoff: bool = False
    parallel_join: bool = False
    bounded_refinement: bool = False
    pause_resume: bool = False
    revision_memory: bool = False
    completion_contract: bool = False

    @property
    def signals(self) -> int:
        return sum((self.typed_handoff, self.parallel_join, self.bounded_refinement,
                    self.pause_resume, self.revision_memory, self.completion_contract))

    @property
    def decision(self) -> str:
        return "evaluate LangGraph" if self.signals >= 2 else "keep a plain pipeline"

profiles = [
    TaskProfile("find a symbol by name"),
    TaskProfile("answer one codebase question", typed_handoff=True),
    TaskProfile("plan a production change", True, True, True, True, True, True),
]
rows = "\n".join(f"| {p.name} | {p.signals} | {p.decision} |" for p in profiles)
display(Markdown("| Task | Signals | Decision |\n|---|---:|---|\n" + rows))
```

The first task has no workflow state beyond a query and result. The last task must preserve evidence across independent searches, revisit gaps, pause for review, and remember validated findings. Those are concrete reasons to evaluate graph orchestration; retrieval quality still belongs to ordinary, measurable search components.

## Baselines before the framework

The first retrieval experiment compares lexical, dense, structural, and hybrid search on the same question. Later chapters add graph control only after this search surface is visible.


```python
from IPython.display import Markdown, display
from change_planner.baselines import compare_control_models


comparison = compare_control_models("retry-01")
rows = "\n".join(
    f"| {row['approach']} | {row['evidence_recall']:.2f} | "
    f"{row['parallel']} | {row['resumable']} |"
    for row in comparison
)
display(Markdown("| Approach | Evidence recall | Parallel | Resumable |\n|---|---:|:---:|:---:|\n" + rows))
assert [row["approach"] for row in comparison] == [
    "direct search",
    "staged Python planner",
    "LangGraph workflow",
]
assert comparison[-1]["parallel"] is True
assert comparison[-1]["resumable"] is True
```

The hybrid result is not automatically correct because it combines more signals. Its value must be established against relevance judgments, latency, and cost. Chapter 02 turns the investigation boundary into typed graph state; Chapter 03 makes each search result recoverable evidence.
