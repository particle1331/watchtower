Planning a change to a production system is not general planning. The relevant evidence is usually inside the system itself: code, tests, configuration, documentation, Git history, and the record of earlier investigations. A useful plan must recover that evidence, relate it to the requested change, expose uncertainty, and remain valid for a specific repository revision.

This course builds a **Change Planner Agent with LangGraph**. The agent answers codebase questions, maps likely impact, connects implementation paths to tests, analyzes regression hypotheses, and produces a cited change plan. Search and retrieval are learned as measurable engineering systems; LangGraph is introduced where investigation state, parallel branches, bounded refinement, durable execution, memory, or human review changes the workflow.

**Design rule:** A change plan may claim only what recoverable, revision-matched evidence supports; deterministic search and analysis own that evidence, while LangGraph owns the stateful decisions around it.


## What you will build

By the end, the course produces one cumulative Change Planner artifact:

- a repository ingestion layer that preserves revision, path, symbol, test, configuration, documentation, and content-fingerprint metadata;
- lexical, dense, hybrid, and reranked retrieval paths compared against fixed relevance judgments;
- typed evidence, relationship, regression-hypothesis, test-link, memory, review, and change-plan objects;
- a LangGraph workflow for scoping, parallel investigation, evidence fusion, targeted verification, review, and export;
- explicit recovery for ambiguous requests, missing evidence, stale indexes, failed tools, incomplete branches, and exhausted search or verification budgets;
- checkpointed investigations plus revision-aware episodic, semantic, and procedural memory; and
- a layered evaluation report covering retrieval, impact analysis, test linkage, regression localization, memory behavior, workflow invariants, latency, and cost.

The completed capstone accepts a bounded change request and a versioned repository snapshot. It emits a Markdown change plan and a machine-readable investigation record containing current behavior, proposed changes, affected surfaces, related tests, regression hypotheses, verification steps, rollout and rollback considerations, observability checks, evidence, and unresolved questions.

The canonical agent is read-only. It may run explicitly allowed tests in the course sandbox, but it does not edit code, merge changes, deploy systems, or treat a repository snapshot as complete knowledge of production runtime state.

## Prerequisites

Tooling:

- Python, a uv-managed environment, and comfort reading typed Python packages.
- Basic command-line and Git skills, including reading a diff and identifying a commit.
- Enough pytest familiarity to run focused tests and interpret failures.
- A local checkout of the repository. No paid API or hosted database is required for the deterministic smoke path.

Assumed knowledge:

- Python functions, classes, type annotations, and simple asynchronous or parallel work.
- Model messages, structured outputs, and tool calls.
- The difference between source evidence, an engineering inference, and an observed test result.

No prior LangGraph, information-retrieval, embedding, reranking, or agent-memory experience is required. The course develops those concepts from measurable baselines before composing them into the agent.


## Execution at a glance

| Path | Environment | Role in the course |
|---|---|---|
| Deterministic smoke | Small pinned Python repositories, committed fixture artifacts, and a scripted planning adapter | Exercises every graph route, evidence contract, seeded regression, and memory policy without network access or paid credentials. |
| Standard local search | The same fixtures with locally rebuilt lexical, dense, hybrid, symbol, test, and history indexes | Produces the canonical retrieval and change-analysis measurements; rebuilding dense and reranking artifacts may require a one-time model download. |
| Real repository integration | A learner-selected local repository and an explicit test-execution policy | Tests portability and surfaces unsupported languages or repository conventions; it does not contribute to the frozen correctness score. |
| Optional live model | An explicitly enabled model API behind the same typed adapter | Exercises live query decomposition and synthesis while preserving deterministic retrieval, evidence validation, and export boundaries. |

The smoke path proves workflow behavior. The standard local path owns the search and retrieval comparisons. Real-repository and live-model runs are integration evidence, not a substitute for the pinned evaluation fixtures.


## Important notes

Reusable computation belongs in the backing `change-planner` package. The notebooks provide the argument: each chapter introduces one retrieval, code-intelligence, graph, memory, or evaluation contract; exercises the shared implementation; and interprets a stored result. They do not become a second implementation of the agent.

Every evidence item preserves repository identity, revision, source kind, recoverable location, and a content fingerprint. Every durable memory also records the investigation and evidence that produced it. A retrieved memory is prior experience, not current repository truth; the agent must revalidate it against the target revision before using it as evidence.

:::{.callout-warning}
The Change Planner is a read-only decision-support system. Repository evidence can reveal code-level impact and missing verification, but it cannot certify production safety or observe runtime facts that were not supplied. A human remains responsible for the final implementation, operational review, rollout, and rollback decisions.
:::

Read [00. Architecture and contracts](00-overview.html) before the numbered chapters. It defines the objects, workflow, memory lifecycle, and success conditions that later chapters assume.
