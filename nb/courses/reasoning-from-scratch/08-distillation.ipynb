categories: [large-language-models, reasoning, distillation]

A reasoning model that answers well may still be too slow or costly for repeated use. This chapter will transfer teacher outputs to a smaller student and test whether the student preserves useful task performance with lower inference cost.

## Planned coverage

- Generate or prepare teacher reasoning examples and define a student training split.
- Fine-tune the smaller model on those examples and inspect the learned response format.
- Compare teacher and student quality, latency, and tokens on the same held-out tasks.
