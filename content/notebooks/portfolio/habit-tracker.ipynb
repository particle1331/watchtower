# Habit Tracker


## The problem

Track daily habits and current streaks from the terminal while keeping records between runs. This small application also serves as a case study of AI-assisted development with explicit architectural boundaries and a reviewable build history.

## What it contains

The Python CLI supports adding habits, recording dated check-ins, listing current streaks, and exporting a versioned JSON document. SQLite stores the records; an interchangeable in-memory repository supports testing.

The implementation separates domain rules, application services, adapters, and the CLI. Architecture tests check dependency and clock rules. Seven build journals record plans, approvals, verification, and review findings.
