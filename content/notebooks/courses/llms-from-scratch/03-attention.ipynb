# Causal Self-Attention and Multi-Head Attention



The text pipeline supplies embeddings; attention lets each position select useful earlier context. This chapter will build that operation in stages.

## Planned content

Derive query, key, and value projections, scaled scores, softmax weights, and weighted values. Explain why the causal mask applies before softmax. Extend a single head to parallel heads and an output projection, tracking tensor shapes.

## Planned lab and evidence

Begin with a tiny numeric calculation, then implement masked single-head and multi-head modules. Check that valid attention rows sum to one, future weights vanish, and changing a future input cannot change an earlier output. Compare explicit arithmetic with the module on a deterministic fixture. The component and its shape contract pass to Chapter 04.
