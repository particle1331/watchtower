# Fine-Tuning GPT for Text Classification



A language model can also predict a discrete label. This chapter will add a classification head and compare its held-out result with a simple baseline.

## Planned content

Study label encoding, class balance, sequence representation, task loss, which parameters to train, and per-class metrics. Keep the task split and checkpoint provenance explicit.

## Planned lab and evidence

Prepare labeled data, inspect class counts, fine-tune from a declared GPT checkpoint, and report a confusion matrix and suitable class metrics. Compare a simple baseline and inspect errors. The classifier result remains separate from language-model validation and becomes one capstone branch.
