# Pretraining, Validation, and Generation



With batches and a complete GPT available, the course can measure learning. This chapter will connect next-token cross-entropy to a training loop and save evidence for the resulting checkpoint.

## Planned content

Explain loss over shifted targets, held-out validation, optimizer updates, overfitting, generation sampling, and checkpoint provenance. Distinguish a locally trained model from externally loaded pretrained weights.

## Planned lab and evidence

Run a tiny smoke experiment and an optional longer run with explicit resource settings. Record train and validation loss, generate from fixed prompts at selected checkpoints, and save model and run configuration together. If weights are loaded, verify parameter mapping and record their source. The declared checkpoint starts Chapters 06–07.
