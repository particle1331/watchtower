# Improve Reasoning with Inference-Time Scaling



Inference-time scaling allocates more generation work to a question without updating model weights. This chapter will compare prompt-based reasoning with strategies that sample several candidates and use agreement or selection to choose an answer.

## Planned coverage

- Compare direct answers and chain-of-thought prompting under the same task and scoring contract.
- Generate multiple candidates and examine self-consistency and best-of-N selection.
- Plot quality against candidate count, generated tokens, and latency to expose the compute tradeoff.
