# Understanding Reasoning Models



Reasoning models are language models that use additional computation or training to improve performance on tasks that require several dependent steps. This opening chapter will define the course's operational meaning of reasoning, distinguish visible reasoning traces from verified correctness, and explain why the project starts from an existing pretrained model.

## Planned coverage

- Map the base model, inference-time methods, reinforcement learning, and distillation.
- Separate this course's scope from building and pretraining a language model.
- Establish the questions the later evaluations should answer: what improved, at what cost, and on which tasks.
