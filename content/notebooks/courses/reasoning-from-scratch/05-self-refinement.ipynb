# Inference-Time Scaling through Self-Refinement



Self-refinement gives a model additional chances to inspect and revise its own answer. This chapter will implement a controlled draft, critique, and revision loop and test whether its extra inference improves answers on the course evaluation set.

## Planned coverage

- Define the information passed between draft, critique, and revision steps.
- Compare one-pass generation with one or more refinement rounds.
- Track correctness, tokens, and latency, including cases where revision does not help.
