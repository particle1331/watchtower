# Train Reasoning Models with Reinforcement Learning



Inference-time methods leave the base model unchanged; reinforcement learning adjusts its weights using a reward signal. This chapter will introduce reinforcement learning from verifiable rewards (RLVR) and trace the small-scale training loop from sampled answers to reward and policy updates.

## Planned coverage

- Define prompts, completions, verifiable rewards, and the policy being optimized.
- Follow the GRPO-based training path in the source repository and identify its inputs and outputs.
- Compare the trained checkpoint with the pretrained baseline using the same evaluation rules.
