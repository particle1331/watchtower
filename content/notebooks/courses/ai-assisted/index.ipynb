AI coding agents can produce code faster than its owner can build a mental model of it. This course follows a small Habit Tracker from pure domain logic to persistence, use cases, a command-line interface, and a deliberately injected architecture violation. The build journals show the plans, mistakes, review findings, and corrections that made each change inspectable.

**Design rule:** The human owns the architecture, and the important rules have executable checks.

## What you will finish with

You will be able to write an architecture brief, scope a change before implementation, enforce a dependency rule, test whether a guardrail detects a violation, and record the evidence for a review. The worked example provides a complete Python CLI with `add`, `checkin`, `list`, and versioned JSON `export`, backed by interchangeable memory and SQLite repositories.

This is a case study with practice tasks. The downloadable project is the final implementation, so the journals describe earlier stages rather than a sequence of incomplete starter checkouts. The example demonstrates the mechanisms on a small codebase; it does not establish time savings or success at large scale.

## How to read the course

Start with the process overview, then follow the ordered course path. Each numbered chapter retains one build journal and adds a practice task. Read the historical plan before inspecting the final code, and use the review and lessons to compare intention with evidence. Historical test counts describe the original build stages.

## Prerequisites

**Knowledge.** Basic Python functions, classes, exceptions, imports, and tests. Familiarity with Git diffs helps when reviewing the record. No machine-learning knowledge is required.

**Tools.** Python 3.9 or later, a terminal, and a text editor. The Habit Tracker and its `unittest` suite use only the standard library. An AI coding agent is optional for trying the workflow; reading and running the case study requires no model API, paid service, or GPU.

## Run the worked example

[Download the worked example](source/ai-assisted-project.zip) and open its `ai-assisted/toy-project/` directory. In the Watchtower checkout, use this equivalent path from the repository root:

```sh
cd projects/ai-assisted/toy-project
python3 -m unittest discover -s tests -t . -v
python3 tools/audit.py
python3 -m habits --help
```

The CLI stores its database in `habits.sqlite3` in the working directory. Set `HABITS_DB` to a scratch database when experimenting, so each exercise can use independent data. The code, tests, and original journals live together under `projects/ai-assisted/`.
