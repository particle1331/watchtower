# Prevent Leakage with Grouped Splits


A test score is informative only when the test examples represent a meaningful challenge. If related source and generated texts appear on both sides of a random row split, a model can exploit shared content and appear to generalize. This chapter turns provenance into a grouped split design.

## Rows are not always independent

Suppose source document $g$ produces a human example and one or more AI rewrites. The examples share topic, named entities, structure, and sometimes phrases. A row-wise split can put one example in training and its sibling in test. The model then sees much of the answer in advance. Instead, assign every row with group key $g$ to one partition.

Choose the split unit from the dependency structure, not from the convenient dataframe shape. If multiple documents belong to one author, source site, or generation batch and that relation matters to the intended claim, a larger group may be appropriate. Grouping by source reduces one leakage path but does not guarantee robustness to new generators or new domains.

## Roles of the partitions

Training data fits model parameters. Validation data supports architecture, hyperparameter, and threshold choices. Test data estimates performance after those decisions are fixed. Repeatedly tuning against test results turns the test set into another validation set. Keep the split manifest and random seed so later runs can reproduce the assignment.

The repository reports grouped train/validation/test splits for its 50k dataset. Confirm the current dataset version and grouping field before relying on that claim; never infer split semantics from filenames alone.
