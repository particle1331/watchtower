# Build a Dataset with Reliable Labels


A classifier can only learn the distinctions represented by its examples and labels. This chapter follows the repository's data-generation stages: licensed human text, AI-generated text, and AI-assisted edits. The goal is to understand provenance and label construction before treating the resulting corpus as a table of trustworthy ground truth.

## Provenance is part of the example

A training row should preserve the text, label, source document or source family, generation or editing method, model/version where relevant, collection time, license, and processing history. The source repository records row-level source/license metadata and keeps raw texts outside Git. Preserve these fields in the data contract even when a training library only consumes text and label; they support grouping, audits, and removal requests.

A human-authored source can be paired with a generated continuation or rewrite. Such a pair is useful because it controls some content differences, but it also creates a shortcut: a model may recognize a topic or source rather than authorship. AI-assisted editing makes the binary boundary less clear still. A minimally corrected human draft and a heavily rewritten draft may have different labels despite sharing an author and source. Record the procedure instead of hiding this ambiguity.

## Dataset checks before modeling

Inspect counts by label, source, domain, length, language, and generation/editing method. Check duplicates and near-duplicates. Sample rows from every source and label combination. Verify that license conditions permit the planned use. Confirm that labels agree with the written labeling rule, then document exceptions rather than silently relabeling them.

Balanced class counts simplify some comparisons but do not establish realistic prevalence. If deployment prevalence differs, precision and the useful operating threshold can change. A benchmark assembled for controlled comparison should be described as such.

The accompanying [classifier-history article](https://magazine.sebastianraschka.com/p/classifier-history-and-jev) makes a practical point about data work: adding carefully labeled examples can improve a task more than spending the same effort on hyperparameter tuning. Treat this as a hypothesis to measure for your task. Chapter 08 uses learning curves to compare data quantity with model and resource choices; better data quality and coverage still matter more than row count alone.
