# Train a TF-IDF Logistic Regression Baseline


A sparse linear classifier provides a quick reference before GPU training or API calls. The repository's logistic-regression stage combines TF–IDF features, a linear decision rule, calibration, and artifact export. This chapter explains why that baseline is both useful and easy to overinterpret.

## From tokens to a score

Term frequency–inverse document frequency (TF–IDF) represents a document with weighted token or character n-gram counts. Logistic regression maps that feature vector to a class score through a linear function followed by a sigmoid. The method is fast and often surprisingly competitive because writing style and formatting leave frequent surface cues.

Fit the vocabulary and inverse-document frequencies on training text only. Applying a vectorizer fit to all partitions leaks information about test vocabulary. Select regularization and any threshold using validation data. Freeze the preprocessing and model together so inference applies the same vocabulary and weighting used during training.

## Interpret the result cautiously

Inspect large positive and negative feature weights alongside representative examples. They can reveal both plausible stylistic cues and collection artifacts such as source names, headers, or templated prompts. A feature weight describes the fitted dataset relationship; it does not identify a causal property of AI authorship.

Compare the baseline with a majority-class predictor and report precision, recall, F1, and a thresholded confusion matrix. Include the threshold and the validation rule that selected it. Save the dataset/split version, preprocessing configuration, classifier parameters, and artifact checksum.
