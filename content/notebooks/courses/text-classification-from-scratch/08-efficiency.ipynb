# Measure Data Efficiency and Inference Cost


Model selection is a resource decision and, for general-purpose systems, a coverage decision. A larger or more flexible model may handle more tasks while increasing cost, latency, or operational dependence. A narrow classifier may be a better fit when one well-defined task repeats at high volume. This chapter makes the tradeoff explicit.

## Learning curves and task breadth

For a task-specific model, train the same configuration on increasing subsets while keeping validation and test data fixed. Repeat subset sampling or report the seed because one small subset can be unusually easy. Track the metric used for the decision. A curve that has not flattened may indicate that more data could help; it does not identify which new examples will improve generalization.

For a general-purpose decision model, map performance across distinct tasks, not just many datasets from one task. Include binary, exclusive multi-class, multi-label, and ordinal outputs where relevant. Report performance and uncertainty by task, and include a task-specific baseline. A wide list of possible applications is not evidence that quality transfers to them.

## Resource accounting

Measure warm and cold start separately, and report hardware, batch size, input length, precision, and concurrency. Distinguish throughput from single-request latency and peak memory from model-file size. For hosted services, include measured price, token or request use, network latency, rate limits, and privacy constraints. For local models, include download size and serving requirements.

Raschka's [article](https://magazine.sebastianraschka.com/p/classifier-history-and-jev) reports a Jev IMDb experiment with high test accuracy and low API cost, and compares it with a ModernBERT run. It also notes repeated-run variation, possible test contamination, and different execution conditions. The lesson is to report those conditions beside the numbers. Do not treat the article's measurements or a vendor benchmark as a universal cost-quality ranking.

## Exercise: cost-quality and coverage frontier

Build a table for the logistic baseline, one transformer, and optionally a prompted LLM or decision API. Include task, held-out split, quality, calibration or threshold metric, training cost, model or API version, model size if known, median and tail latency, and memory or API cost. Mark values that are unavailable rather than inventing estimates.

Identify which system is dominated for your intended use. If you claim one model is more general, show results on at least two distinct tasks and explain the remaining benchmark, population, and data-overlap limits. A cheaper classifier with different error behavior is not automatically a better fit.
