# Advanced Topics

These notebooks apply the foundations from the [Deep Learning Foundations](../index.html) series to real tasks and architectures. Each notebook is self-contained, uses a small publicly available dataset, and is designed to run on a single GPU — with CPU-accessible paths provided where training is expensive.

## About This Series

**Audience.** Someone who has worked through the core Deep Learning Foundations notebooks (01–09) and wants to see the same building blocks — attention, encoder-decoder architectures, convolutional networks — assembled into complete, working systems on real datasets.

**Stack.** PyTorch throughout. Datasets sourced from `torchtext`, `torchvision`, and Hugging Face `datasets`. No high-level training frameworks. Every design choice is traced back to the underlying math covered in the prerequisites.

**Prerequisites.** [NB09](../09-attention-transformers.html) (attention and Transformers) is required for NB13 and NB17. [NB06](../06-cnn.html) and [NB12](../12-appendix-resnet.html) are required for NB16. NB14 and NB15 need only [NB06](../06-cnn.html).

## How to Read This Series

**If you want sequence modeling:** Start with [NB13](./13-translation.html) — it introduces the full encoder-decoder Transformer and is a prerequisite for NB17.

**If you want generative models:** Read [NB14](./14-generative-models.html) (VAE + GAN) then [NB15](./15-diffusion.html) (DDPM/DDIM). They share the same dataset and build naturally on each other.

**If you want dense prediction:** [NB16](./16-segmentation.html) (segmentation) is self-contained after NB06 and NB12.

**If you want cross-modal generation:** [NB17](./17-captioning.html) (captioning) builds directly on NB13's decoder and ViT from NB09.
