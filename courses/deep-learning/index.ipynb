## About This Series

This series builds the mathematical and computational foundations of deep learning from scratch — in pure Python and PyTorch, deriving everything from first principles.

**Audience.** Someone comfortable with Python, NumPy, and basic calculus who wants to understand how neural networks actually work before moving on to language models. A mathematics background is assumed — derivations are shown in full, not waved away. Code is the other half of the explanation.

**Stack.** Pure Python and PyTorch. No high-level training frameworks. Every concept is grounded in code you write yourself. The core series ([01](01-softmax-regression.ipynb)–[09](09-attention-transformers.ipynb)) covers the full story from a single linear classifier through multi-layer networks, automatic differentiation, modern optimizers, CNNs, RNNs, and the Transformer encoder. Advanced Topics (NB13–17) apply the foundations to [real tasks and architectures]{.mark}. This provides a minimal prerequisite foundation for the [Language Models from Scratch](../llm/index.ipynb) series.

## Prerequisites

- Python 3.10+, PyTorch 2.0+, NumPy, Matplotlib
- Linear algebra (matrix products, eigenvalues), calculus (chain rule, partial derivatives). Derivations are shown in full.
- Most notebooks run on CPU. [Notebook 06](06-cnn.ipynb) (CNNs) includes ResNet fine-tuning on a real image dataset — a GPU (e.g. T4) is strongly recommended for that section.

## How to Read This Series

**If you're new to deep learning:** Work through [01](./01-softmax-regression.html) → [05](./05-activations-and-gradients.html) in order — they build directly on each other. On a first pass through 05, focus on the results rather than every experiment; the code is extensive.

**If you want the full picture:** Continue with [06](./06-cnn.html) → [07](./07-language-modeling.html) → [08](./08-rnn.html) → [09](./09-attention-transformers.html) to go from CNNs through language models, RNNs, and Transformers. The appendices are optional enrichment — [NB11](./11-appendix-expressivity.html) for approximation theory, [NB12](./12-appendix-resnet.html) for CNN architecture depth, [NB18](./18-appendix-double-descent.html) for the double descent phenomenon and why the classical bias-variance story is incomplete.

**If you want applied practice:** After 09, work through the [Advanced Topics](./advanced/index.html) (13–17) to see the foundations applied to translation, generative models, diffusion, segmentation, and captioning.

**Next steps:** After the core notebooks, continue with [Language Models from Scratch](../llm/index.html).
