This course covers the foundations of modern reinforcement learning — from
MDPs and dynamic programming through policy gradients, PPO, and RL for
language models.

## Chapters

| # | Chapter | Topics |
|---|---------|--------|
| 1 | [RL Foundations](01-rl-foundations.ipynb) | MDPs, value functions, Bellman equations, dynamic programming, TD learning, Q-learning, function approximation |
| 2 | [Policy Gradients](02-policy-gradients.ipynb) | REINFORCE, baselines, variance reduction, actor-critic methods |
| 3 | [Proximal Policy Optimization](03-ppo.ipynb) | GAE, trust regions, PPO-Clip, implementation, diagnostics |
| 4 | [RL for Language Models](04-rl-for-llms.ipynb) | RLHF, DPO, PPO for LMs, GRPO, reward hacking, KL constraint |
