# Classical Machine Learning

Despite the rise of deep learning, [classical ML methods remain dominant in industry]{.mark} for structured and tabular data — the form most enterprise data actually takes. Tree-based ensembles consistently outperform neural networks on tabular benchmarks, logistic regression ships in production systems where interpretability is required, and the statistical reasoning underlying these methods — bias-variance tradeoff, calibration, causal inference — is precisely [what separates engineers who can build models from engineers who can trust them]{.mark}. This series reviews the classical toolkit at the depth expected of a senior practitioner.

## About This Series

A concise but rigorous review of classical machine learning for engineers with deep learning backgrounds who want to fill gaps in the statistical and tabular ML foundations.

**Audience.** Senior engineers comfortable with PyTorch and deep learning fundamentals who want a grounded treatment of classical methods, statistical reasoning, and the practical tradeoffs that govern model selection in production systems.

**Stack.** [scikit-learn](https://scikit-learn.org), [XGBoost](https://xgboost.readthedocs.io), [LightGBM](https://lightgbm.readthedocs.io), [SHAP](https://shap.readthedocs.io), [Optuna](https://optuna.readthedocs.io), and PyTorch for the tabular deep learning notebook. Each notebook includes runnable experiments on small publicly available datasets.

## Prerequisites

- **Python 3.10+**, NumPy, Matplotlib, scikit-learn
- **Basic probability and statistics:** distributions, expectations, Bayes' rule
- **Familiarity with gradient descent** and loss minimization at the level of [Deep Learning Foundations NB01](/courses/deep-learning/01-softmax-regression.html)

## How to Read This Series

**If you want the statistical foundations:** Start with [NB01](/courses/classical-ml/01-bias-variance.html) and [NB05](/courses/classical-ml/05-evaluation-causal.html) — these cover the reasoning framework (bias-variance, calibration, causal inference) that underpins sound ML practice regardless of which model class you use.

**If you work primarily with tabular data:** Prioritize [NB02](/courses/classical-ml/02-tree-methods.html) (tree methods dominate tabular benchmarks) and [NB04](/courses/classical-ml/04-feature-engineering.html) (feature engineering determines most of the performance ceiling).

**If you want to understand when deep learning applies to tables:** [NB06](/courses/classical-ml/06-tabular-dl.html) directly addresses this question with benchmarks across model classes.

**If you're coming from Deep Learning Foundations:** NB01 connects to the double descent appendix ([NB18](/courses/deep-learning/18-appendix-double-descent.html)), which shows how the classical bias-variance story breaks down in the overparameterized regime.
