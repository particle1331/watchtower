# Agentic Systems

An AI agent is an LLM equipped with tools, memory, and a control loop that lets it pursue goals across multiple steps — perceiving its environment, deciding what to do next, acting, and updating based on the result. Most developers encounter agents through frameworks that abstract away the mechanics. This series goes the other direction: it builds the picture from first principles, starting with the foundational papers that defined the field and ending with multi-agent orchestration and runtime extensibility via the Model Context Protocol.

## About This Series

This is a technical deep-dive into the [design of agentic systems]{.mark} — not a tutorial on building applications. Every notebook has a runnable experiment that makes the concept concrete and measurable. The goal is to build intuition that transfers to new problems, not familiarity with a specific framework.

**Audience.** Someone who has built or closely read an LLM-powered application and wants to understand *why* the design choices work, when they break down, and what the alternatives are. Comfort with the OpenAI Python SDK and basic async Python is assumed.

**Stack.** All notebooks use a custom library (`src/notebooks/agent/`) called **CDA** (Coding Agent), built from scratch in the companion [Coding Agent From Scratch](/courses/app-dev/cda/index.html) series in `apps/cda/`. The library is deliberately small and readable — every mechanism we discuss here has a corresponding implementation you can read in a few hundred lines. Notebooks that focus on patterns (NB00, NB08) use raw `AsyncOpenAI` calls to keep the pattern logic visible.

## Prerequisites

- Python 3.13+, comfort with `async`/`await`, basic OpenAI SDK usage
- No prior agent framework experience required
- NB00 is self-contained; NB01 onward assume NB00 or equivalent background

## How to Read This Series

**If you are new to agents:** Start at [NB00](/courses/agents/00-foundations.html) for the conceptual foundations, then follow in order through NB03. After NB03 you understand the full agent loop; the remaining notebooks each add one mechanism at a time.

**If you already understand tool calling:** Start at [NB03](/courses/agents/03-agent-loop.html) to see how the CDA library implements the loop end-to-end, then jump to whatever mechanism interests you — context (NB04), memory (NB05), safety (NB06), or evaluation (NB07).

**If you want practical patterns:** [NB08](/courses/agents/08-patterns.html) and [NB09](/courses/agents/09-multi-agent.html) are relatively self-contained. NB08 covers four structured workflow patterns from Anthropic's "Building Effective Agents"; NB09 extends them to multi-agent coordination.

**If you are here for MCP:** [NB10](/courses/agents/10-extensibility.html) requires only NB01 (tool calling) and NB03 (agent loop) as background. It shows how the CDA library connects to any MCP server — local subprocess or remote HTTP — without modifying library code.
