A reasoning method cannot be judged from a compelling sample alone. This chapter will create a small, reproducible evaluation path that checks answers against task criteria and makes parsing errors and benchmark limits visible.

## Planned coverage

- Define a task slice and its expected-answer format before running comparisons.
- Extract final answers and apply exact-match or task-specific checks.
- Summarize accuracy, inspect representative failures, and state what the evaluation cannot establish.
