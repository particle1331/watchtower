A decoder-only language model predicts the next token from the tokens already seen. The course will follow one small GPT through data preparation, model construction, pretraining, and two downstream adaptations. This overview defines what each stage changes and how its result will be judged.

## Running system

Text becomes token IDs. Fixed-length input windows and one-token-shifted targets form batches. Token and position embeddings enter decoder blocks that combine causal self-attention with feed-forward networks. A language-model head produces vocabulary scores. Generation chooses a next token and appends it to the context.

Causal masking enforces the information boundary: an output position can use earlier positions and itself, but cannot use future tokens. The attention and model chapters will test this before training.

| Stage | Artifact | Evidence |
|:--|:--|:--|
| Text | Tokenizer interface and batches | Decode round trip, shapes, shifted targets |
| Model | Forward pass and generation | Causal checks, shapes, parameter count |
| Pretraining | Checkpoint and run record | Train and validation loss, fixed-prompt samples |
| Classification | Task checkpoint | Held-out confusion matrix and class metrics |
| Instructions | Task checkpoint | Fixed-prompt comparisons and failure notes |
| Optional LoRA | Adapter weights | Trainable-parameter count and matched-task result |

## Execution and evaluation

The default path uses a small configuration and short local run. It demonstrates that interfaces and optimization work, but cannot establish broad language ability. An optional substantial run may use a GPU, more data, longer training, or externally supplied pretrained weights. These results must be reported separately because their compute and starting points differ.

PyTorch supplies tensor operations, automatic differentiation, and optimizers. The course owns model composition, data contracts, checks, and interpretation. A tokenizer package or downloaded weights may be used when its dependency and provenance are recorded.

Split data before fitting or model selection. Record tokenizer, context length, corpus subset, seed, optimizer, step count, device, and checkpoint identity. Compare pretraining loss on held-out sequences and generate from fixed prompts. For classification, report class balance and held-out per-class metrics. For instruction tuning, review fixed prompts for compliance, correctness, and unsupported claims. Example generations are qualitative evidence, not a general capability score.

Chapters 01–02 establish the task and batches; 03–04 build the decoder; 05 trains and saves it; 06–07 branch into task objectives; 08 optionally changes the update method; and 09 joins the run records and limitations. The [source repository](https://github.com/rasbt/LLMs-from-scratch) provides the core Chapters 1–7 and LoRA in Appendix E.
