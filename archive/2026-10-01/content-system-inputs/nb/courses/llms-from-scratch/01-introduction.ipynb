This chapter will make the goal of building an LLM a specific next-token modeling task. Its decisions set the interfaces and comparison rules used later.

## Planned content

Explain how a sequence prefix becomes an input and its following token becomes a target. Distinguish architecture, pretraining, generation, and downstream fine-tuning. Contrast an untrained or simple frequency baseline with the trained model, and state what a small educational run can demonstrate.

## Planned lab and evidence

Define a run record with model configuration, tokenizer, data split, seed, device, optimizer, and checkpoint identifier. Choose a tiny local configuration and fixed generation prompts. Trace a short sequence into input-target pairs and explain why future-token leakage invalidates the result. The experiment contract passes to Chapter 02.
