This optional extension revisits a downstream task while changing how weights are updated. It will test trainable-parameter cost and task quality under a matched setup.

## Planned content

Explain a frozen weight matrix plus a low-rank update, adapter rank and placement, and fair comparison against full fine-tuning on the same split and starting checkpoint.

## Planned lab and evidence

Attach adapters to selected projections, verify that base weights remain frozen, and count trainable parameters. Repeat the classification or instruction task with the same evaluation procedure. Report resource cost and measured quality without assuming either method wins. Include adapter results in the capstone only if the lab is completed.
