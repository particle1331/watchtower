A text classifier maps an input text to a label, a score, or both. This course uses the repository's AI-written-text detector to make that mapping concrete, but the same distinctions apply to sentiment, spam, topic, and moderation classifiers. Before looking at architectures, define what counts as a labeled example and what action a score is meant to support.

## Task contract

Let $x$ be a text and $y\in\{0,1\}$ its recorded label, where this course uses $1$ for “AI-generated under the dataset's labeling procedure” and $0$ for “human-written under that procedure.” The wording matters: the label records how an example entered the dataset. It is not a direct observation of authorship in every possible sense.

A model estimates a score $s(x)$. A decision rule applies a threshold $t$ to produce $\hat y$. Changing $t$ changes the number of texts flagged, even when the model is unchanged. A system specification therefore includes intended users, the positive class, the costs of each error, input limits, and how a person can contest or interpret an output.

## Why this is a classification course

The source project deliberately compares a conventional sparse-feature baseline, a prompted-model baseline, and fine-tuned encoder and decoder models. That sequence separates two questions: whether a method learns predictive signals at all, and whether a more complex model improves the evidence enough to justify added cost and operational risk. “From scratch” here means following the end-to-end experiment—from labels and data through evaluation and serving—not implementing every neural network primitive from first principles.

A detector may correlate with the training distribution without detecting authorship as a general property. A generator, editor, domain, or language shift can change the cues. False positives can harm human writers; false negatives can miss generated content. Treating the score as an accusation would claim more than a supervised classifier demonstrates.

## First experiment contract

Before running a model, write down: the exact dataset version; the row and group identifiers; the split; which label is positive; which metric selects a model; which metric selects a threshold; and the intended use. Chapter 02 audits where labels came from, and Chapter 03 turns the group identifier into a leakage-resistant split plan.

## Exercises

### Exercise 1: Make the label operational

Write a two-sentence labeling rule for the binary task. Include how AI-assisted editing is handled, and identify one case the rule cannot classify cleanly.

### Exercise 2: Choose an error tradeoff

Imagine a writing platform deciding whether a post should receive a voluntary review notice. Explain what could go wrong after a false positive and after a false negative. State which error should weigh more for that use and why. Do not assume the same threshold fits a school, hiring process, or spam filter.

### Exercise 3: Separate score from claim

Rewrite “This text is 82% AI-written” as a statement that accurately describes a calibrated model score and its dependence on an evaluation distribution.
