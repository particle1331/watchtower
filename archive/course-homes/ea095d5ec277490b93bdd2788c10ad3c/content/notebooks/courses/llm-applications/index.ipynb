# Build LLM Applications: Retrieval, Agents, and Evaluation


## About this course

Teach the application mechanisms around a pretrained language model: retrieval, tool execution, context selection, structured responses, and evaluation. Build small components whose contribution can be measured against a simple baseline. Model architecture, pretraining, and reinforcement-learning fine-tuning belong to the existing companion courses.

## Who this is for

Python programmers familiar with functions, classes, JSON, tests, and basic model inference. The AI-assisted development course supplies architecture and review habits; Build LLMs from Scratch supplies tokenizer and generation intuition. Neither completed pretraining nor a GPU is required. Early labs use local fixtures and recorded model responses; a small open model or hosted API is an optional measured extension.

## Learning outcomes

Implement a bounded tool-calling loop; score complete agent runs; compare lexical, dense, hybrid, and reranked retrieval; assemble permission-filtered evidence; validate structured answers and citations; compare prompt versions and iterative retrieval on fixed cases; report correctness, abstention, latency, tokens, and cost separately.

## The running project

A small document assistant over synthetic support notes and technical documents. Each chapter supplies a bounded fixture and can be implemented independently. An optional cumulative implementation combines ingestion, search, generation, and read-only tools behind explicit interfaces. Keep deterministic offline checks separate from live model experiments. Service deployment is developed in Deploy and Operate AI Systems.

## Practice and assessment

Every lab records its reproducing command, fixture and model versions, baseline, measured result, and one limitation. The final report separates retrieval misses, unsupported claims, wrong answers, unnecessary abstentions, permission leaks, and resource use. Compare changes on frozen development/test cases; reserve stronger statistical claims for results with suitable repeated runs and uncertainty estimates. Improvements, regressions, and inconclusive comparisons are all valid findings.
